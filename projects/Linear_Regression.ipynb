### A] To predict housing prizes based on features such as size and location

#### Step 1: import libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
```

#### Step 2: Load Dataset

```python
df = pd.read_csv(r"C:\Users\swaya\01_College\Sem_05\FML\House Price Prediction Dataset - House Price Prediction Dataset.csv")
print(f"Original Dataset:\n{df.head()}")
```

#### Step 3: Dataset information:

```python
print(f"Shape of dataset:\n{df.shape}")
```

```python
print(f"\nColumns:\n{df.columns}")
```

```python
print(f"\nMissing Values:\n{df.isnull().sum()}")
```

#### Step 4: Select `feature (X)` and `target (y)`

```python
X = df[["Area"]]
y = df["Price"]
```

#### Step 5: Split Dataset

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.20, random_state = 42)
```

#### Step 6: Create Linear Regression model

```python
model = LinearRegression()
```

#### Step 7: Train model

```python
model.fit(X_train, y_train)
print("model trained successfully")
```

#### Step 8: Predict test data

```python
y_pred = model.predict(X_test)
print(f"Actual prices:\n{y_test.values}")
```

```python
print(f"Predicted prices:\n{y_pred}")
```

#### Step 9: Model Equation

```python
print("Slope (Coefficient)")
print(model.coef_[0])
```

```python
print("Intercept")
print(model.intercept_)
```

#### Step 10: Evaluation metrics

```python
mae = mean_absolute_error(y_test, y_pred)

mse = mean_squared_error(y_test, y_pred)

rmse = np.sqrt(mse)

r2 = r2_score(y_test, y_pred)

print(f"Mean Absolute Error: \t\t{mae}")
print(f"Mean Squared Error: \t\t{mse}")
print(f"Root Mean Squared Error: \t{rmse}")
print(f"R square: \t\t\t{r2}")
```

#### Step 11: Predict new house prize

```python
new_house = pd.DataFrame({"Area":[1750]})

predicted_price = model.predict(new_house)
print(f"Predicted prize for 1750 sq.ft house:\n{predicted_price[0]}")
```

#### Step 12: Visualization

```python
plt.figure(figsize = (8,6))
plt.scatter(X, y, label = "Actual Data")
plt.plot(X, model.predict(X), linewidth=2, color='Black', label='Regression Line')

plt.title("Linear Regression - House Price Prediction : Swayam-16")
plt.xlabel("House size(sq.ft.)")
plt.ylabel("House Price")

plt.legend()
plt.grid(True)
plt.show()
```
